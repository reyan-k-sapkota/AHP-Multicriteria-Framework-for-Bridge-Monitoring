```python
structural_indicators = [
    "Material degradation",
    "Foundation/scour vulnerability",
    "Structural deformation",
    "Joint condition",
    "Deck integrity",
    "Live load response",
    "Environmental exposure",
]
#done
strategic_indicators = [
    "Traffic volume & network criticality",
    "Redundancy; availability of alternate routes",
    "Socioeconomic importance of served area",
    "Role in disaster response / emergency access",
]

#done
def pair_order(indicators):
    """Index pairs in questionnaire order: (0,1),(0,2),...,(n-2,n-1)."""
    n = len(indicators)
    return [(i, j) for i in range(n) for j in range(i + 1, n)]

#done
structural_pairs = pair_order(structural_indicators)   # 21 pairs
#done
strategic_pairs  = pair_order(strategic_indicators)    #  6 pairs

#done
print(f"There are in total {len(structural_pairs)} structural and {len(strategic_pairs)} strategic pairwise questions.")
```

```python
def tally_table(tally, item_a, item_b):
    """Readable table for one question's tally."""
    t = pd.DataFrame({"Answer value": list(tally.keys()), "Experts": list(tally.values())})
    t = t.sort_values("Answer value", ascending=False).reset_index(drop=True)
    t["Share of experts"] = (t["Experts"] / t["Experts"].sum()).round(3)
    t["Meaning"] = t["Answer value"].apply(
        lambda v: f"'{item_a}' and '{item_b}' equally important" if v == 1 else
                  (f"'{item_a}' more important (x{v:g})" if v > 1 else f"'{item_b}' more important (x{1/v:.0f})"))
    return t

q1 = MANUAL_TALLIES["structural"][(0, 1)]
print("Question 1:", structural_indicators[0], "vs", structural_indicators[1])
display(tally_table(q1, structural_indicators[0], structural_indicators[1]))
print(f"Total experts for this question: {sum(q1.values())}")


#similarly, lets see tally for Material degradation vs Structural deformation
q2 = MANUAL_TALLIES["structural"][(0, 2)]
print("Question 2:", structural_indicators[0], "vs", structural_indicators[2])
display(tally_table(q2, structural_indicators[0], structural_indicators[2]))
print(f"Total experts for this question: {sum(q2.values())}")


#similarly, lets see tally for Material degradation vs Joint condition
q3 = MANUAL_TALLIES["structural"][(0, 3)]
print("Question 3:", structural_indicators[0], "vs", structural_indicators[3])
display(tally_table(q3, structural_indicators[0], structural_indicators[3]))
print(f"Total experts for this question: {sum(q3.values())}")

#similarly, you can see tally for other questions as well.
```

```python
#Done
# Overview of all 28 questions in the survey data
overview = []
def summarise(block, indicators, pair_list, category):
    for (i, j) in pair_list:
        t = block[(i, j)]
        total = sum(t.values())
        first = sum(c for v, c in t.items() if v > 1)
        second = sum(c for v, c in t.items() if v < 1)
        equal = sum(c for v, c in t.items() if v == 1)
        overview.append({"Block": category, "Item A": indicators[i], "Item B": indicators[j], "Experts": total,
                         "A favoured": first, "Equal": equal, "B favoured": second,
                         "Most common answer": max(t.items(), key=lambda kv: kv[1])[0]})

summarise(MANUAL_TALLIES["structural"], structural_indicators, pair_order(structural_indicators), "Structural")
summarise(MANUAL_TALLIES["strategic"], strategic_indicators, pair_order(strategic_indicators), "Strategic")
o = MANUAL_TALLIES["overall"]
overview.append({"Block": "Overall", "Item A": "Structural", "Item B": "Strategic", "Experts": sum(o.values()),
                 "A favoured": sum(c for v, c in o.items() if v > 1), "Equal": o.get(1.0, 0),
                 "B favoured": sum(c for v, c in o.items() if v < 1),
                 "Most common answer": max(o.items(), key=lambda kv: kv[1])[0]})
overview_df = pd.DataFrame(overview)
overview_df.index = overview_df.index + 1
display(overview_df)
print(f"All questions have {overview_df['Experts'].nunique()} distinct respondent total(s): "
      f"{sorted(overview_df['Experts'].unique())}")
```

```python
# Step 1: turn ONE question's tally into ONE number
# Question: Material degradation vs Foundation/scour vulnerability  -> pair (0, 1)
#using geomtric mean method

tally = MANUAL_TALLIES["structural"][(0, 1)] #for Question: Material degradation vs Foundation/scour vulnerability  -> pair (0, 1)

values = np.array(list(tally.keys()), dtype=float)    # the answers given: 9, 8, 7, ... 0.333
counts = np.array(list(tally.values()), dtype=float)  # how many experts gave each answer

print("values:", values)
print("counts:", counts)
print("total experts:", counts.sum())

# a) log of each value
log_values = np.log(values)

# b) each log counted once per expert who gave that answer
weighted_logs = counts * log_values

# c) add up, divide by number of experts, undo the log
mean_log = weighted_logs.sum() / counts.sum()
group_value = np.exp(mean_log)

print("\nsum of (count x ln value):", round(weighted_logs.sum(), 4))
print("mean log:", round(mean_log, 4))
print("group value (geometric mean):", round(group_value, 4))
```

```python
# Step 2: put ONE group value into a 7 x 7 matrix (with its reciprocal)

n = len(structural_indicators)          # 7 indicators
A = np.ones((n, n))                     # start with all 1s (diagonal is already correct)

i, j = 0, 1                             # Material degradation (0) vs Foundation/scour (1)
A[i, j] = group_value                   # row 0, column 1
A[j, i] = 1 / group_value               # mirror cell = reciprocal

# short codes so the matrix fits on screen: S1 ... S7
codes = [f"S{k+1}" for k in range(n)]
print("S1 = Material degradation, S2 = Foundation/scour vulnerability, ...\n")

display(pd.DataFrame(A, index=codes, columns=codes).round(4))
```

```python
# Step 3: fill the full 7 x 7 structural matrix, one pair at a time

n = len(structural_indicators)
A_struct = np.ones((n, n))                       # start with all 1s

for (i, j) in structural_pairs:                  # (0,1), (0,2), ... (5,6)
    tally  = MANUAL_TALLIES["structural"][(i, j)]
    values = np.array(list(tally.keys()),   dtype=float)
    counts = np.array(list(tally.values()), dtype=float)

    gm = np.exp(np.sum(counts * np.log(values)) / counts.sum())   # as done in Step 1

    A_struct[i, j] = gm                          # Step 2
    A_struct[j, i] = 1 / gm

    print(f"{structural_indicators[i]:32s} vs {structural_indicators[j]:32s} -> {gm:.4f}")

codes = [f"S{k+1}" for k in range(n)]
print("\nStructural group matrix:")
display(pd.DataFrame(A_struct, index=codes, columns=codes).round(4))

# sanity check: every cell times its mirror cell must equal 1
print("Reciprocal check (A * A.T all equal to 1):", np.allclose(A_struct * A_struct.T, 1))
```

```python
# Step 4: weights from the principal eigenvector of the structural matrix (done by Eigenvector method)

#note: Sir's paper uses the column-normalization method, not the eigenvector method. 
# It is the quick approximation and it is simpler than this method

"""
Which one to use?
Both are accepted. The eigenvector method is Saaty's original and the theoretically exact one. 
The column-normalization method is easier to explain and to do by hand, and many papers use it, as this one does. 
If it follows Saaty, use the eigenvector.
"""

# 1) all eigenvalues and eigenvectors
eigvals, eigvecs = np.linalg.eig(A_struct)
print("All eigenvalues:")
print(np.round(eigvals, 4))

# 2) position of the largest eigenvalue (compare real parts only)
idx = np.argmax(eigvals.real)
lambda_max = eigvals[idx].real
print(f"\nLargest eigenvalue (lambda_max) = {lambda_max:.4f}   (n = {n})")

# 3) its eigenvector -> positive numbers -> divide by the sum
raw_vector = eigvecs[:, idx].real
print("\nRaw eigenvector (before cleaning):", np.round(raw_vector, 4))

w = np.abs(raw_vector)
weights_struct = w / w.sum()

result = pd.DataFrame({"Indicator": structural_indicators, "Weight": weights_struct}, index=codes)
display(result.round(4))
print("Sum of weights:", round(weights_struct.sum(), 4))
```

```python
#by column-normalization method

# Step 4 (sir's paper's method): column-normalise, then average each row

col_sums = A_struct.sum(axis=0)                 # sum of each column
print("Column sums:", np.round(col_sums, 4))

N = A_struct / col_sums                         # Eq. (2): each cell / its column sum
display(pd.DataFrame(N, index=codes, columns=codes).round(4))
print("Each column of N sums to:", np.round(N.sum(axis=0), 4))

weights_paper = N.mean(axis=1)                  # Eq. (3): average of each row
result = pd.DataFrame({"Indicator": structural_indicators, "Weight": weights_paper}, index=codes)
display(result.round(4))
print("Sum of weights:", round(weights_paper.sum(), 4))


print("Notice how sir's method and AHP's eigen vector method give very similar results. The eigenvector method is theoretically exact, but the column-normalization method is easier to explain and to do by hand.")
print("Since The eigenvector method is Saaty's original and the theoretically exact one, we will use this method for our work.")
```

```python
# Step 4: weights from the principal eigenvector of the structural matrix

# 1) all eigenvalues and eigenvectors
eigvals, eigvecs = np.linalg.eig(A_struct)
print("All eigenvalues:")
print(np.round(eigvals, 4))

# 2) position of the largest eigenvalue (compare real parts only)
idx = np.argmax(eigvals.real)
lambda_max = eigvals[idx].real
print(f"\nLargest eigenvalue (lambda_max) = {lambda_max:.4f}   (n = {n})")

# 3) its eigenvector -> positive numbers -> divide by the sum
raw_vector = eigvecs[:, idx].real
print("\nRaw eigenvector (before cleaning):", np.round(raw_vector, 4))

w = np.abs(raw_vector)
weights_struct = w / w.sum()

result = pd.DataFrame({"Indicator": structural_indicators, "Weight": weights_struct}, index=codes)
display(result.round(4))
print("Sum of weights:", round(weights_struct.sum(), 4))
```

```python
# Step 5: consistency check of the structural matrix

# Saaty's Random Index for matrices of size n
RI_table = {1: 0.00, 2: 0.00, 3: 0.58, 4: 0.90, 5: 1.12,
           6: 1.24, 7: 1.32, 8: 1.41, 9: 1.45, 10: 1.49} #these are a bit different in some sources. Verify with Arun sir

CI = (lambda_max - n) / (n - 1)
RI = RI_table[n]
CR = CI / RI

print(f"lambda_max = {lambda_max:.4f}")
print(f"n          = {n}")
print(f"CI = (lambda_max - n) / (n - 1) = ({lambda_max:.4f} - {n}) / ({n} - 1) = {CI:.4f}")
print(f"RI (Saaty table, n = {n}) = {RI}")
print(f"CR = CI / RI = {CI:.4f} / {RI} = {CR:.4f}")

if CR < 0.10:
    print("\nCR < 0.10  ->  the group judgments are acceptably consistent.")
else:
    print("\nCR >= 0.10 ->  the judgments are too inconsistent; review the inputs.")
```

```python
# Step 6: the same steps as a reusable function, applied to the STRATEGIC block

RI_table = {1: 0.00, 2: 0.00, 3: 0.58, 4: 0.90, 5: 1.12,
            6: 1.24, 7: 1.32, 8: 1.41, 9: 1.45, 10: 1.49}

def ahp_block(tallies, indicators, pairs):
    #this function takes the tallies, indicators and pairs as input and returns the AHP matrix (calculated as Geometric Means), weights (by Eigen-vector method), lambda_max (it is the max eigen value), CI and CR
    n = len(indicators)

    # Steps 1-3: tally -> geometric mean -> matrix (with reciprocals)
    A = np.ones((n, n))
    #building AHP matric
    for (i, j) in pairs:
        t = tallies[(i, j)]
        values = np.array(list(t.keys()),   dtype=float)
        counts = np.array(list(t.values()), dtype=float)
        gm = np.exp(np.sum(counts * np.log(values)) / counts.sum())
        A[i, j], A[j, i] = gm, 1 / gm

    # Step 4: weights from the principal eigenvector
    #findings weights using eigen vector method
    eigvals, eigvecs = np.linalg.eig(A)
    idx = np.argmax(eigvals.real)
    lam = eigvals[idx].real
    w = np.abs(eigvecs[:, idx].real)
    w = w / w.sum()

    # Step 5: consistency check
    CI = (lam - n) / (n - 1)
    CR = CI / RI_table[n]
    return A, w, lam, CI, CR

A_strat, weights_strat, lam_strat, CI_strat, CR_strat = ahp_block(
    MANUAL_TALLIES["strategic"], strategic_indicators, strategic_pairs)

g_codes = [f"G{k+1}" for k in range(len(strategic_indicators))]
print("Strategic group matrix:")
display(pd.DataFrame(A_strat, index=g_codes, columns=g_codes).round(4))

display(pd.DataFrame({"Indicator": strategic_indicators, "Weight": weights_strat}, index=g_codes).round(4))
print(f"lambda_max = {lam_strat:.4f}, n = 4, CI = {CI_strat:.4f}, CR = {CR_strat:.4f}")

# bonus check: the function must reproduce your structural result
_, w_check, _, _, CR_check = ahp_block(MANUAL_TALLIES["structural"], structural_indicators, structural_pairs)
print("\nStructural check -> CR:", round(CR_check, 4), "| weights match Step 4:", np.allclose(w_check, weights_struct))
```

```python
# Step 7: category weights (Structural vs Strategic)

overall_tally = MANUAL_TALLIES["overall"]
values = np.array(list(overall_tally.keys()),   dtype=float)
counts = np.array(list(overall_tally.values()), dtype=float)
print("Total experts:", counts.sum())

# weighted geometric mean (same as Step 1)
r = np.exp(np.sum(counts * np.log(values)) / counts.sum())
print(f"r = {r:.4f}  -> Structural is judged about {r:.2f} times as important as Strategic")

# 2 x 2 matrix, shown for clarity
A_cat = np.array([[1, r],
                  [1 / r, 1]])
display(pd.DataFrame(A_cat, index=["Structural", "Strategic"], columns=["Structural", "Strategic"]).round(4))

# exact eigenvector solution for a 2 x 2 reciprocal matrix
#Those lines are a shortcut for the eigenvector calculation, for a 2×2 matrix, and can be written as a simple formula.
w_structural = r / (1 + r)
w_strategic  = 1 / (1 + r)
print(f"\nw_structural = r / (1 + r) = {r:.4f} / {1 + r:.4f} = {w_structural:.4f}")
print(f"w_strategic  = 1 / (1 + r) = 1 / {1 + r:.4f} = {w_strategic:.4f}")
print("Sum:", round(w_structural + w_strategic, 4))

# optional check: does np.linalg.eig give the same answer?
ev, evec = np.linalg.eig(A_cat)
vec = np.abs(evec[:, np.argmax(ev.real)].real)
print("Eigenvector check:", np.round(vec / vec.sum(), 4))
```

```python
# Step 8: global weights = local weight x category weight

global_struct = weights_struct * w_structural     # 7 structural indicators
global_strat  = weights_strat  * w_strategic      # 4 strategic indicators

print("Check - structural global weights add up to the category weight:",
      round(global_struct.sum(), 4), "vs", round(w_structural, 4))
print("Check - strategic global weights add up to the category weight:",
      round(global_strat.sum(), 4), "vs", round(w_strategic, 4))

# put all 11 indicators in one table
global_df = pd.DataFrame({
    "Category":        ["Structural"] * 7 + ["Strategic"] * 4,
    "Indicator":       structural_indicators + strategic_indicators,
    "Local weight":    np.concatenate([weights_struct, weights_strat]),
    "Category weight": [w_structural] * 7 + [w_strategic] * 4,
    "Global weight":   np.concatenate([global_struct, global_strat]),
})

# sort from most to least important and add rank
global_df = global_df.sort_values("Global weight", ascending=False).reset_index(drop=True)
global_df["Rank"] = global_df.index + 1

display(global_df[["Rank", "Category", "Indicator", "Local weight", "Global weight"]].round(4))
print("Sum of all 11 global weights:", round(global_df["Global weight"].sum(), 4))
```
